```python

destinations = (
    (201, "Ooty", "Hill Station", 4000),
    (202, "Goa", "Beach", 5000),
    (203, "Mysore", "Heritage", 3500),
    (204, "Munnar", "Hill Station", 7000),
    (205, "Manali", "Mountain", 6000),
    (206, "Jaipur", "Heritage", 4500),
    (207, "Kerala", "Backwaters", 5500),
    (208, "Kashmir", "Valley", 8000)
)
class Destination:
    def __init__(self, code, name, category, fare):
        self.code = code
        self.name = name
        self.category = category
        self._fare = fare
        self.__state = "AVAILABLE"
    def reserve(self):
        self.__state = "BOOKED"
    def state(self):
        return self.__state
    def cost_for(self, travelers):
        return self._fare * travelers
class TravelAgency:
    def __init__(self):
        self.destination_list = {}
        self.confirmed = []
    def add_destination(self, destination):
        self.destination_list[destination.code] = destination
    def display_destinations(self):
        print("\n========== TRAVEL DESTINATION MENU ==========")
        for destination in self.destination_list.values():
            print(
                destination.code,
                "-",
                destination.name,
                "-",
                destination.category,
                "- Rs." + str(destination._fare)
            )
    def book(self, code, travelers):
        if code not in self.destination_list:
            print("Invalid Destination Code.")
            return
        if travelers <= 0:
            print("Number of travelers must be greater than zero.")
            return
        destination = self.destination_list[code]
        if destination.state() == "BOOKED":
            print("This destination is already booked.")
            return
        basic_amount = destination.cost_for(travelers)
        discount_rate = lambda n: 0.10 if n >= 4 else (0.05 if n >= 2 else 0)
        discount = basic_amount * discount_rate(travelers)
        payable = basic_amount - discount
        destination.reserve()
        record = (
            destination.name,
            travelers,
            payable
        )
        self.confirmed.append(record)
        print("\n========== BOOKING DETAILS ==========")
        print("Destination:", destination.name)
        print("Category:", destination.category)
        print("Number of Travelers:", travelers)
        print("Trip Cost: Rs." + str(basic_amount))
        print("Discount: Rs." + str(discount))
        print("Final Cost: Rs." + str(payable))
        print("Booking Status:", destination.state())
    def display_confirmed(self):
        print("\n========== CONFIRMED BOOKINGS ==========")

        if not self.confirmed:
            print("No confirmed bookings.")
            return
        for record in self.confirmed:
            print(
                record[0],
                "-",
                record[1],
                "Travelers - Rs." + str(record[2])
            )
agency = TravelAgency()
for details in destinations:
    agency.add_destination(
        Destination(
            details[0],
            details[1],
            details[2],
            details[3]
        )
    )
agency.display_destinations()
try:
    code = int(input("\nEnter Destination Code: "))
    travelers = int(input("Enter Number of Travelers: "))
    agency.book(code, travelers)
except ValueError:
    print("Please enter valid numeric values.")
agency.display_confirmed()
```
